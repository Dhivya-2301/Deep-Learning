```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import SGD, RMSprop, Adam
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train / 255.0
x_test = x_test / 255.0

optimizers = {
    "SGD": SGD(),
    "RMSProp": RMSprop(),
    "Adam": Adam()
}

history_data = {}

for name, optimizer in optimizers.items():

    model = Sequential([
        Flatten(input_shape=(28, 28)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    print(f"\nTraining with {name}...\n")

    history = model.fit(
        x_train,
        y_train,
        epochs=5,
        batch_size=32,
        verbose=1
    )

    history_data[name] = history.history

print("\nFinal Training Results")
print("-" * 40)

best_optimizer = ""
best_accuracy = 0

for name in history_data:
    loss = history_data[name]['loss'][-1]
    accuracy = history_data[name]['accuracy'][-1]

    print(f"{name}")
    print(f"Loss     : {loss:.4f}")
    print(f"Accuracy : {accuracy:.4f}\n")

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_optimizer = name

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)

for name in history_data:
    plt.plot(history_data[name]['loss'], label=name)

plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.subplot(1,2,2)

for name in history_data:
    plt.plot(history_data[name]['accuracy'], label=name)

plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.tight_layout()
plt.show()

print("Performance Comparison")
print("-" * 40)

for name in history_data:
    print(f"{name} Accuracy = {history_data[name]['accuracy'][-1]:.4f}")

print(f"\nBest Optimizer: {best_optimizer}")
```
