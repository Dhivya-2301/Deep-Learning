```python
import numpy as np
np.set_printoptions(precision=3,suppress=True)
x=np.array([
    [1.0, 2.0, 30.0],
    [2.0, 4.0, 32.0],
    [3.0, 6.0, 28.0],
    [4.0, 8.0, 34.0],
])
print("Input data: \n",x)
```

```python
eps=1e-5
mean=x.mean(axis=0)
var=x.var(axis=0)
print("Mean:",mean)
print("Variance: ",var)
x_hat=(x-mean/np.sqrt(var+eps))
print("Normalized x \n",x_hat)
print("Mean per Feature: ",x_hat.mean(axis=0))
print("Variance per Feature: ",x_hat.var(axis=0))
```

```python
gamma=np.ones(3)
beta=np.zeros(3)
out=gamma*x_hat+beta
print("Output after gamma/beta: \n",out)
```

```python
mean_ln = x.mean(axis=1, keepdims=True)
var_ln = x.var(axis=1, keepdims=True)

x_hat_ln = (x-mean_ln) / np.sqrt (var_ln + eps)
print("LayerNorm normalized x:\n", x_hat_ln)
print("\nrow means(should be ~0):", x_hat_ln.mean(axis=1))
print("row vars (should be ~1):", x_hat_ln.var(axis=1))
```
